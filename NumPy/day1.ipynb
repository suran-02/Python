## AGGREGATE FUNCTIONS

```python
import numpy as np

arr = np.array([[31,20,30,44,53,12],
                  [69,37,48,90,10,11]])

print(np.sum(arr))
print(np.mean(arr))
print(np.std(arr))
print(np.var(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmax(arr))
print(np.argmin(arr))
print(np.sum(arr,axis=0)) # sum the columns
print(np.sum(arr,axis=1))   # sum the rows
```
