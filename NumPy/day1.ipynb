```python
### NUMPY ARRAYS & BASICS
import numpy as np

## creating arrays
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_1d = np.array(list1)
arr_1d.ndim
arr_2d = np.array([list1,list2])
arr_2d.ndim
arr_3d = np.array([[list1,list2,list3]])
arr_3d.ndim
```

```python
## LIST vs NUMPY ARRAYS
py_list = [1,2,3]
py_list*3
arr_1d*3
```

## NUMPY ARRAY ATTRIBUTES

```python
print(arr_2d.shape)
print(arr_2d.size)
print(arr_2d.dtype)
print(arr_2d.ndim)
```

## ARRAY INITIALIZATION METHODS

```python
#zeros_arrays
zero_arr = np.zeros((2,3))
zero_arr
```

```python
#ones_Arrays
one_arr = np.ones((2,3))
one_arr
```

```python
#full arrys
full_arr = np.full((2,3),5)
full_arr
```

```python
#Identity Matrix
id_arr = np.eye(5)
id_arr
```

```python
#Empty Arrays
print(np.empty(2))
```

```python
# Evenly spaced array
np.arange(1,10,2)
```

```python
# specific number of equally (spaced) values between a range
np.linspace(1,10,4)
```

```python
# random values array (float)

r_Arr = np.random.rand(3,2)
r_Arr
```

```python
# random values array (int)
rint_Arr = np.random.randint(1,1000,(3,2))
rint_Arr
```
