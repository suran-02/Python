## FILTERING

```python
import numpy as np

ages = np.array([[21, 17, 19, 20, 16, 30, 18, 65],
                 [39, 22, 14, 99, 18, 19, 20, 21]])

teens = ages[ages < 18]
# adults = ages[(ages >= 18) & (ages < 65)]

# boolean indexing
seniors = ages[ages >= 65]
# print(seniors)
evens = ages[ages % 2 == 0]
# print(evens)

# preserves original shape of the array
adults = np.where(ages>=18,ages,0)
print(adults)
```
