```python
import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles

#pyplot is a collection of functions that make matplotlib work as MATLAB

#1 Create first chart

#data
x=[1,2,3,4,5]
y=[10,20,25,30,50]
plt.plot(x,y) #line chart
plt.show() #display chart in jpnb
```

## CUSTOMIZE CHART

```python
# import matplotlib.pyplot as plt
plt.figure(figsize=(4,3)) #figure size
plt.plot(x,y,color="#d93d76",marker='o',linestyle='--',linewidth=2,markersize=12)

plt.title("Line Chart")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

## ADVANCED LINE CHART - MULTIPLE LINES AND LEGENDS

```python
plt.figure(figsize=(4,3))
x1=[10,20,30,40,50]
y1=[15,21,33,41,57]
y2=[23,34,48,55,62]

plt.plot(x1,y1,label="Sales 2025")
plt.plot(x1,y2,label="Sales 2026")
plt.title("YoY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

## BAR CHART

```python
x1=['A','B','C','D','E']
y1=[15,21,80,41,57]
plt.bar(x1,y1)
plt.title("Bar Chart")
plt.show()
```

## HISTOGRAM

```python
#used for distribution analysis

import random
data = [random.randint(1,10) for _ in range(100)]
plt.hist(data,bins=10)
plt.title("Histogram")
plt.show()
```

## PIE CHART

```python
# used to show part-to-whole relationship

x1=['A','B','C','D','E']
y1=[15,21,80,41,57]

plt.pie(y1, labels=x1,autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```
