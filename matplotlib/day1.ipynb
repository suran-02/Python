```python
import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles

#pyplot is a collection of functions that make matplotlib work as MATLAB

#1 Create first chart

#data
x=[1,2,3,4,5]
y=[10,20,25,30,50]
plt.plot(x,y) #line chart
plt.show() #display chart in jpnb
```

## CUSTOMIZE CHART

```python
# import matplotlib.pyplot as plt
plt.figure(figsize=(4,3)) #figure size
plt.plot(x,y,color="#d93d76",marker='o',linestyle='--',linewidth=2,markersize=12)

plt.title("Line Chart")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

## ADVANCED LINE CHART - MULTIPLE LINES AND LEGENDS

```python
plt.figure(figsize=(4,3))
x1=[10,20,30,40,50]
y1=[15,21,33,41,57]
y2=[23,34,48,55,62]

plt.plot(x1,y1,label="Sales 2025")
plt.plot(x1,y2,label="Sales 2026")
plt.title("YoY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
